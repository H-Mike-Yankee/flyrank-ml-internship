## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*


**Chosen lane: Lane 2 — Refresh / Content Opportunity Scoring**

**ML task type: Ranking**

This problem is best framed as a ranking task because the goal is not only to predict whether a page needs attention or not. The goal is to prioritize thousands of content pages and identify which pages should be reviewed first.

The model output will support a ranked list of pages based on their refresh opportunity score.

A higher-ranked page represents a stronger candidate for content review, expansion, or improvement.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


The ideal target would be the future impact of refreshing a page, such as whether a content update improves performance.

However, that outcome is not available in the starter dataset because we do not observe future refresh results.

Therefore, I will use a proxy target based on observed page performance signals and define refresh opportunity using existing metrics that indicate potential improvement areas.

The proxy represents a decision-support signal, not a guaranteed prediction of future traffic.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric is Precision@K.

This measures how many of the top-ranked pages selected by the model are actually relevant according to the defined evaluation target.

For this task, a higher Precision@50 means the model is better at identifying useful pages for the first review batch.

A good model should outperform the fixed baseline rule.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


A fixed rule can identify simple cases, but content performance depends on multiple interacting signals.

For example, a page may have high impressions but low clicks, declining engagement, or strong ranking potential.

These patterns are difficult to capture with a single if-statement.

A machine learning model can learn combinations of signals and produce a more flexible ranking that supports better prioritization decisions.
